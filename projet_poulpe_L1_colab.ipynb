# Rappels de mécanique

## Deuxième loi de Newton

La dynamique du système est gouvernée par :

$$
m \frac{dv}{dt} = \sum F
$$

où :
- $m$ est la masse,
- $v$ la vitesse,
- $\sum F$ la somme des forces.

---

## Poussée d’Archimède

Un corps plongé dans un fluide subit une force verticale :

$$
F_A = \rho g V
$$

où :
- $\rho$ est la masse volumique du fluide,
- $V$ le volume déplacé.
